```python
import matplotlib.pyplot as plt

#显示中文
plt.rcParams['font.sans-serif'] = ['PingFang SC']  # Mac常用中文字体
plt.rcParams['axes.unicode_minus'] = False  # 正常显示负号

#1. 创建子图
#figure 画布对象, axes:子图数组
figure, axes = plt.subplots(nrows=2, ncols=2, figsize=(20, 6), dpi=100)  # 创建2行2列的子图 元组tuple解包,

'''
nrows 表示行数
ncols 表示列数
figsize 表示画布大小
dpi 表示图片清晰度

axes[0,0] 表示第1行第1列的子图
axes[0,1] 表示第1行第2列的子图
axes[1,0] 表示第2行第1列的子图
axes[1,1] 表示第2行第2列的子图

注意：索引从0开始，不是从1开始
'''

#2. 准备数据
x = [1, 2, 3, 4, 5]
y1 = [10, 12, 15, 13, 18]
y2 = [5, 8, 6, 10, 12]
y3 = [20, 18, 22, 25, 24]
y4 = [100, 200, 150, 300, 250]

#3. 在不同子图中绘图
axes[0, 0].plot(x, y1)  # 在第1行第1列画折线图
axes[0, 0].set_title('axes[0,0] 折线图')  # 设置子图标题

axes[0, 1].bar(x, y2)  # 在第1行第2列画柱状图
axes[0, 1].set_title('axes[0,1] 柱状图')  # 设置子图标题

axes[1, 0].scatter(x, y3)  # 在第2行第1列画散点图
axes[1, 0].set_title('axes[1,0] 散点图')  # 设置子图标题

axes[1, 1].plot(x, y4)  # 在第2行第2列画折线图
axes[1, 1].set_title('axes[1,1] 折线图')  # 设置子图标题

#4. 调整布局并显示
plt.tight_layout()  # 自动调整子图间距，防止标题重叠
plt.show()  # 显示图表
```

```python
from matplotlib.axes import Axes

figure1, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 6), dpi=100)

#图一:柱状图(世界石油储备)--bar方法
countries = ['China','America','Russia','Singapore','Canada','England','Australia']
values = [35,23,24,6,15,12,18]

axes1:Axes = axes[0]
axes1.bar(countries, values,width=0.7)
axes1.set_title('世界石油储备',fontsize=18)
axes1.set_xlabel('Country',fontsize=18)
axes1.set_ylabel('Petroleum reserves',fontsize=18)
axes1.grid(linestyle='--',alpha=0.3)

#在Jupyter notebook中 会默认执行plt.show图表展示,python文件中需手动输入

#图二:饼状图(世界人口)--》擅长比例分析--pie方法
countries2 = ['印度', '中国', '美国', '印尼', '巴基斯坦', '尼日利亚', '巴西', '俄罗斯', '其他']
values2 = [14.51, 14.099, 3.4, 2.83, 2.51, 2.33, 2.12, 1.44, 20]

axes2:Axes = axes[1]
axes2.pie(values2, labels=countries2, autopct='%1.1f%%')#autopct百分比格式:
# %1.1f浮点数格式：1是最小宽度，.1是保留 1 位小数
# %%转义符，输出一个真正的 % 百分号（因为单个 % 是格式占位符，要打两个才显示一个）
axes2.set_title('世界人口', fontsize=18)
axes2.legend(loc='lower center', ncol=5,bbox_to_anchor=(0.5, -0.05))#显示图例,ncol:每行显示的个数;bbox_to_anchor:图例位置(0.5控制水平方位,-0.05控制垂直方位)

#保存图片
plt.savefig('data/01.png')

import os
save_path = os.path.abspath('data/01.png')  # 获取图片完整保存路径
plt.savefig(save_path, dpi=300, bbox_inches='tight')  # 保存图片
print(save_path)  # 打印图片保存位置
```
