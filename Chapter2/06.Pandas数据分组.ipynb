```python
from pandas.core.methods.to_dict import to_dict

"""数据分组:把数据按照某个特征分成不同的组,然后对每个组分别进行统计计算
常见方法:
分组后求和: df.groupby('特征')['需要统计的值').sum()
分组后统计数量: df.groupby('特征')['需要统计的值').count()
分组后求最大值: df.groupby('特征')['需要统计的值').max()
分组后求最小值: df.groupby('特征')['需要统计的值').max()
分组后求平均值: df.groupby('特征')['需要统计的值').mean()
聚合方法:df.groupby('特征')['需要统计的值').agg(['sum','count','max','min','mean'])
"""
import pandas as pd

#1.读取数据
df = pd.read_csv('data/sales02.csv',nrows=20)
df['销售金额']=df['单价']*df['销售数量']

#2.分组

#2.1 根据 产品类别 分组,统计各个类别的 订单数量 -count
df.groupby('产品类别').count()#统计时缺失值不参与统计--》选无缺失值的列
```

```python
#2.2 根据 产品类别 分组,统计各个类别的 销售数量之和 -sum
df.groupby('产品类别')['销售数量'].sum()
```

```python
#2.3 根据 产品类别 分组,统计各个类别的 销售金额 之和 -sum
df.groupby('产品类别')['销售金额'].sum()
```

```python
#2.4 根据 产品类别 分组,统计各个类别的最低商品 单价 -min
df.groupby('产品类别')['单价'].min()
```

```python
#2.5 根据 产品类别 分组,统计各个类别的最高商品 单价 -max
df.groupby('产品类别')['单价'].max()
```

```python
#2.6 根据 产品类别 分组,统计各个类别的平均商品 单价 -mean
df.groupby('产品类别')['单价'].mean()
```

```python
#2.7 根据 产品类别 分组,统计各个类别的商品 平均单价,最高单价,最低单价
df.groupby('产品类别')['单价'].agg(['mean','max','min'])
df.groupby('产品类别').agg  ({'单价':['mean','max','min']})
```

```python
#2.8 根据 产品类别 分组,统计各个类别的商品的 销售数量之和,销售金额之和,平均单价
df.groupby('产品类别').agg  ({'销售数量':'sum','销售金额':'sum','单价':'mean'})
```

```python
df.groupby('产品类别').agg  ({'销售数量':'sum','销售金额':'sum','单价':'mean'}).to_dict() #转字典
```
