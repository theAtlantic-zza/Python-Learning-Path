```python
import pandas as pd

#1.读取数据
df = pd.read_csv('data/sales02.csv',nrows=10)

#2.排序
"""数据排序时,一般是升序/降序两种排列方式
升序:df.sort_values(by=['列名'],ascending=True)
降序:df.sort_values(by=['列名'],ascending=False)
Pandas支持多列排序,需分别指定ascending的值,多列排序时.第一个值一样,再排第二个
"""

#2.1 根据 销售数量 倒序排序
df.sort_values('销售数量',ascending=False)
```

```python
#2.2 根据 单价 升序排序
df.sort_values('单价',ascending=True)
df.sort_values('单价') #Default:ascending=True
```

```python
#2.3 根据 单价 升序排序,价格一样,再根据 销售数量 倒序排序
df.sort_values(['单价','销售数量'],ascending=[True,False])
```
